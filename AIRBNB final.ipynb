## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Load your dataset
# Example: df = pd.read_csv('path of csv file')
#creating a copy of dataframe
dataset=df.copy()
````

````python
# Step 1: Check for duplicates and remove them
dataset=dataset.drop_duplicates()
dataset.count()

````

````python
# Step 2: Handle missing values
# Check missing value
dataset.isnull().sum()
````

````python
dataset = dataset.copy()
````

````python
dataset['name'].fillna('Unnamed_Listing', inplace = True)
dataset['host_name'].fillna('Unnamed_Host', inplace = True)

'''
data.fillna({name: Unnamed_Listing}, inplace=True)
data.fillna({host_name: Unnamed_Host}, inplace=True)
'''
````

````python
# calculating the percent of missing values in the last_review & reviews_per_month
missing_count_last_review = dataset['last_review'].isnull().sum()
total_count = len(dataset)

misisng_percent_last_review = (missing_count_last_review / total_count) * 100
print(f"The percentage of missing values in 'last_review' is: {misisng_percent_last_review:.2f}%")

missing_count_review_per_month = dataset['reviews_per_month'].isnull().sum()
total_count = len(dataset)
misisng_percent_reviews_per_month  = (missing_count_last_review / total_count) * 100
print(f"The percentage of missing values in 'reviews_per_month ' is: {misisng_percent_reviews_per_month :.2f}%")
````

````python
# Keep nulls
dataset['last_review'] = pd.to_datetime(dataset['last_review'], errors='coerce')

# Add meaning
dataset['has_review'] = dataset['last_review'].notna()
````

````python
# Fill missing values in 'reviews_per_month' with 0 (assuming no reviews mean zero activity)
dataset.fillna({'reviews_per_month': "0"},inplace=True) # Replace missing  reviews_per_month with "Unknown"
````

#### Chart - 8    Nested pie chart of neighbourhood and neighbourhood groups using nested pie chart

````python
# Grouping data
neighbourhood_group_data = dataset.groupby(['neighbourhood_group', 'neighbourhood'])['number_of_reviews'].sum()
top_neighbourhood_group_data = neighbourhood_group_data.sort_values(ascending=False).head(20)

neighbourhood_group_totals = neighbourhood_group_data.groupby('neighbourhood_group').sum()

# Data for inner pie chart
neighbourhood_totals = top_neighbourhood_group_data.values

# Outer labels for neighbourhood groups
outer_labels = neighbourhood_group_totals.index

# Inner labels for top neighborhoods
inner_labels = [neighbourhood for _, neighbourhood in top_neighbourhood_group_data.index]

colors = plt.cm.tab20.colors

# Plotting
fig, ax = plt.subplots(figsize=(10, 8))

# Outer ring
ax.pie(
    neighbourhood_group_totals,
    labels=outer_labels,
    radius=1,
    autopct='%1.1f%%',
    pctdistance=1.0,
    colors=colors[:len(neighbourhood_group_totals)],
)

# Inner ring
inner_colors = colors[:len(neighbourhood_totals)]
ax.pie(
    neighbourhood_totals,
    labels=inner_labels,
    radius=0.7,
    colors=inner_colors,
    labeldistance=1,
    textprops={'fontsize': 5},
)

# Center circle
centre_circle = plt.Circle((0, 0), 0.5, color='white')
plt.gcf().gca().add_artist(centre_circle)

# Title
plt.title('Nested Pie Chart: Neighbourhood Groups and Neighbourhood')
plt.show()
````

##### 1. Why did you pick the specific chart?

The nested pie chart was chosen because it effectively represent the relationship between neighbourhood groups and the neighbourhoods within them.It visually breaks down the proportions of listings contributed by each neighbourhood in a way that is easy to interpet

##### 2. What is/are the insight(s) found from the chart?

The chart clearly shows which neighbourhoods contribute the most listing with in each group,highlighting drivers for activity .

It reveals whether listings are concentrated in a few neighbourhoods or evenly distributed across the group.

The relationship between the inner and outer ring helps identify how much each neighbourhood contributes to its overall group.
